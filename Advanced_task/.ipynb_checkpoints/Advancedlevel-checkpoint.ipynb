```python
#library 
import pandas as pd
import matplotlib.pyplot as plt
file = pd.read_csv("netflix_titles.csv")
```

```python
#dataset header part 
file.head()
```

```python
file.info()
```

```python
file.shape
```

```python
file.isnull().sum()
```

```python
#filling missing values 
file['country'] = file['country'].fillna('Unknown')
```

```python
file['rating']= file['rating'].fillna('Unknown')
```

```python
file['duration']= file['duration'].fillna('Unknown')
```

```python
#drop rows wiht missing date_added
file = file.dropna(subset =['date_added'])
```

```python
#remove a spces in column date_added
file['date_added'] = file['date_added'].str.strip()
```

```python
file['date_added'] = pd.to_datetime(file['date_added'],errors='coerce')
```

```python
file['date_added'].head()
```

```python
#check  a null value 
file.isnull().sum()
```

```python
file = file.dropna(subset=['date_added'])
file['year_added'] = file['date_added'].dt.year
```

```python
file[['date_added', 'year_added']].head()
```

```python
#What is overall distribution of moviea nd tv show on netflix
import matplotlib.pyplot as plt

type_counts = file['type'].value_counts()

plt.figure(figsize =(6,4))
type_counts.plot(kind = 'bar')

plt.title("Distribution of Movie and TV shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

#Conculsion

From this graph, we can understand that movies are showcased more on Netflix
than TV shows. This means the proportion of movies on Netflix is higher
compared to TV shows.

```python
# How many movies and TV shows were added each year, and which years had the highest number of added?

content_by_year = (
    file.groupby(["year_added","type"]).size().reset_index(name="count")
)
content_by_year.head()
```

```python
import plotly.express as px
fig = px.line(
    content_by_year,
    x = "year_added",
    y = "count",
    color = "type",
    title = "Number of Movies and Tv shows are added on netflix",
    labels = {
        'year_added' : "Year added to netflix",
         'count' : "Number of titles",
         'Type' : "Content Type" 
    }
)
fig.show()
```

#Conclusion

From this graph, we can see that the highest number of movies were added to
Netflix in the year 2019, with approximately 1,424 movies added.
In contrast, the highest number of TV shows were added in 2020, with
around 595 TV shows added.

```python
#How many movies and TV shows were released each year, and which years had the highest number of releases?

released_by_year = (
    file.groupby(['release_year', 'type']).size().reset_index(name ='count')
)
released_by_year.head()
```

```python
#How many movies and TV shows were released each year, and which years had the highest number of releases?

import plotly.express as px

fig = px.line(
    released_by_year,
    x = 'release_year',
    y ='count',
    color = 'type',
    labels = {
        'release_year' :'Release Year',
        'count' :'Number of Titles',
        'type' : 'Content type'
    },
    title = "Number of Movie and Tv shows released each year"
)

fig.show()
```

#Conculsion

The graph shows that most movies added to Netflix were released during 2017 and 2018. In comparison, most TV shows added to Netflix were released 
around the year 2020.

```python
released_by_year.sort_values('count', ascending = False)
```

```python
total_by_year = content_by_year.groupby('year_added')['count'].sum().reset_index(name ='total')
content_prop = content_by_year.merge(total_by_year, on='year_added')
content_prop['proportion'] = content_prop['count'] / content_prop['total']
content_prop.head()
```

```python
#How has the proportion of movies versus TV shows on Netflix changed over time?

import plotly.express as px

fig = px.area(
    content_prop,
    x='year_added',
    y='proportion',
    color='type',
    labels={
        'year_added': 'Year Added to Netflix',
        'proportion': 'Proportion of Content',
        'type': 'Content Type'
    },
    title="Proportion of Movies vs TV Shows Added to Netflix Over Time"
)

fig.show()
```

#Conclusion

The graph shows that movies are added more to Netflix than TV shows in most years.
TV shows increase over time, but movies are still higher overall.

```python
file['gap'] = file['year_added'] - file ['release_year']
file[['title', 'release_year', 'year_added', 'gap']].head()
```

```python
#What is the time gap between a title’s release year and the year it was added to Netflix?

import plotly.express as px

fig = px.violin(
    file,
    x ='type',
    y = 'gap',
    box = True,
    points = 'all',
    color = 'type',
    labels = {
        'gap' :'Gap between relase and added',
        'type' : 'Content Type'
    },
    title = "Distribution of time gap between release and netflix addition by content type."
)
fig.show()
    
```

#Conclusion

This graph shows the difference in time between the release year of content and the year it was added to Netflix. Most movies and TV shows are added
to Netflix within a few years of their release.

However, TV shows show a slightly wider spread, meaning some TV shows take longer to be added to Netflix compared to movies.

```python
file_genres = file [['year_added', 'listed_in']].dropna()
file_genres = file_genres.assign(listed_in = file_genres['listed_in'].str.split(',')).explode('listed_in')
file_genres = file_genres[file_genres['year_added'] >= 2010] 

genre_count = file_genres.groupby(['year_added','listed_in']).size().reset_index(name= 'count')
```

```python
#Which genres have experienced the most growth on Netflix over the last decade?

import plotly.express as px

fig = px.line(
    genre_count,
    x = 'year_added',
    y = 'count',
    color='listed_in',
    title = 'Growth of Netflix genres Over the last decade',
    labels = {
        'year_added' :'year_added',
        'count' : 'Number of titles',
        'listed_in' :'Genre'
    }
    
)

fig.show()
```

#Conclusion

This graph shows the growth of differnt genres over the year. Among all categoried, International Movies show the highest growth 
This indicated that Neetflic has incresedd tis focus on international content form different countreis in recent years.

```python
#explore countries
file_exp = file.assign(country = file['country'].str.split(', ')).explode('country')

country_year = file_exp.groupby(['year_added', 'country']).size().reset_index(name='count')


country_order = country_year.groupby('country')['count'].sum().sort_values(ascending=False).index
```

```python
#Which countries produce the most Netflix content, and how has their contribution changed over time?

fig = px.area(
    country_year,
    x = 'year_added',
    y = 'count',
    color = 'country',
    category_orders= {
        'country' : country_order 
    },
    title ="Netflix content by country over time"
)
fig.update_layout(
    xaxis_title='Year',
    yaxis_title='Number of Titles',
    legend_title='Country',
  
)
fig.show()
```

#Conclusion

This graph shows the growth of Netflix content from different countries over time.
Content additions increased significantly after 2016.
The United States contributes the highest number of titles, followed by
countries such as India and the United Kingdom.
This shows that Netflix has expanded its content globally.

```python
file.head()
```

```python
file['date_added'] = pd.to_datetime (file["date_added"], errors ='coerce')
file['year_added'] = file['date_added'].dt.year
file.columns
```

```python
file['listed_in'] = file['listed_in'].fillna('Unknown')
file_genre = file.assign (genre= file['listed_in'].str.split(', ')).explode('genre')

country_genre_year = (
    file_genre.groupby(['country', 'genre','year_added']).size().reset_index(name = 'count')
)
```

```python
top_countries = (
    file.groupby('country').size().sort_values(ascending = False).head(10).index
)

filtered_data = country_genre_year[
    country_genre_year['country'].isin(top_countries)
]
```

```python
#How do genre trends vary by country over time, and which country–genre combinations show significant growth?
import plotly.express as px

fig = px.line(
    filtered_data,
    x = 'year_added',
    y = 'count',
    facet_col = 'country',
    facet_col_wrap = 2,
    title = 'Genre Trends by Country over time ',
    markers = True
)

fig.update_layout(
    xaxis_title = 'Year',
    yaxis_title = 'Number of Titles',
    legend_title = 'Genre',
    height = 700
    
)
fig.show()
```

#Conclusion

Genre trends change over time and are different for each country.
The United States shows strong growth across many genres.
India and South Korea also show noticeable growth in some genres.
This shows that Netflix focuses on different content types in different countries.

This analysis is based on available data and may be limited by missing country or genre information. In the future, more detailed analysis could
include user viewing data or advanced statistical methods.
